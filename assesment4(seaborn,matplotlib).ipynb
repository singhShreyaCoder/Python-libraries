```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_excel('/content/assignment 3.xlsx')

```

```python
#question 1
import pandas as pd
import matplotlib.pyplot as plt


ms= df.groupby("Month")["Sales"].sum()


# Plot
plt.figure(figsize=(8,5))
plt.plot(ms.index, ms.values,
         color="blue", marker="o", linewidth=2)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

```python
#question 2
rp=df.groupby('Region')['Profit'].sum()
ax=rp.plot(kind='bar',color=['r','g','b','orange'])

for p in ax.patches:
  ax.annotate(f'{p.get_height():0f}',(p.get_x()+0.1,p.get_height()))

plt.show()
```

```python
# QUESTION 3.Create a pie chart showing sales contribution by category.
cs=df.groupby('Category')['Sales'].sum()
exp=[0.1 if v==cs.max() else 0 for v in cs]
plt.pie(cs,labels=cs.index,autopct='%1.1f%%',explode=exp)
plt.title('Sales Contribution by Category')

```

```python
#Create a scatter plot between sales and profit
plt.scatter(df.Sales,df.Profit,s=df.Quantity*40,c=df.Discount,cmap='viridis')
plt.colorbar(label='Discount')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.title('Sales vs Profit')
plt.show()
```

```python
#Create 2×2 subplot layout containing:Line Plot Bar Plot Pie Chart Scatter Plot
fig,axs=plt.subplots(2,2,figsize=(10,8))
axs=axs.ravel()
axs[0].plot(ms.index.astype(str),ms.values,marker=0)
rp.plot(kind='bar',ax=axs[1])
axs[2].pie(cs,labels=cs.index)
axs[3].scatter(df.Sales,df.Profit)
fig.suptitle('dashboard')
plt.tight_layout()
plt.show()
```

```python
#Question 6 Create a countplot showing number of orders from each region.
ax=sns.countplot(data=df,x='Region',hue='Region',legend=False,palette='Set2')
for c in ax.containers:
  ax.bar_label(c)
plt.show()
```

```python
#Question 7 Create a boxplot showing Sales distribution for each Category.
sns.boxplot(data=df,x='Category',y='Sales')
plt.show()
```

```python
#Question 8 Create a violin plot comparing Profit by Gender.
sns.violinplot(data=df,x='Gender',y='Profit')
plt.show()
```

```python
#Question 9 Create a heatmap of numerical variable correlations
# Select only numerical columns for correlation calculation
numerical_df = df.select_dtypes(include=['number'])
corr = numerical_df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
```

```python
#Question 10 Create a pairplot
sns.pairplot(df)
plt.show()


```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_excel('/content/assignment 3.xlsx')

#Question 12 You are a Data Analyst preparing a presentation for management.
#Using only visualizations created above, answer the following:

#Which region performs best?
#north region
rp=df.groupby('Region')['Profit'].sum()
ax=rp.plot(kind='bar',color=['r','g','b','orange'])

for p in ax.patches:
  ax.annotate(f'{p.get_height():0f}',(p.get_x()+0.1,p.get_height()))

plt.show()

 #Which category generates highest revenue?(Electronics)

cs=df.groupby('Category')['Sales'].sum()
exp=[0.1 if v==cs.max() else 0 for v in cs]
plt.pie(cs,labels=cs.index,autopct='%1.1f%%',explode=exp)
plt.title('Sales Contribution by Category')

 #Which month has highest sales?(July)

ms= df.groupby("Month")["Sales"].sum()

plt.figure(figsize=(8,5))
plt.plot(ms.index, ms.values,
         color="blue", marker="o", linewidth=2)

plt.title(" Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()

#Does discount affect profit?(higher discount lower the profit)

plt.scatter(df.Sales,df.Profit,s=df.Quantity*40,c=df.Discount,cmap='viridis')
plt.colorbar(label='Discount')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.title('Sales vs Profit')
plt.show()


```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid", palette="Set2")


# KPI Calculations

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
avg_discount = df["Discount"].mean()
total_orders = df["Order_ID"].count()

# Sales by Region
region_sales = df.groupby("Region")["Sales"].sum()

# Monthly Sales
monthly_sales = df.groupby("Month")["Sales"].sum()
month_order = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug"]
monthly_sales = monthly_sales.reindex(month_order)



# Sales by Category
category_sales = df.groupby("Category")["Sales"].sum()


# Create Dashboard

fig = plt.figure(figsize=(16,10))
fig.suptitle("SALES PERFORMANCE DASHBOARD",
             fontsize=20,
             fontweight="bold")


# KPI Text Boxes

plt.figtext(0.08,0.90,
            f"Total Sales\n₹{total_sales:,.0f}",
            fontsize=12,
            bbox=dict(facecolor="skyblue",
                      edgecolor="black",
                      boxstyle="round,pad=0.5"))

plt.figtext(0.30,0.90,
            f"Total Profit\n₹{total_profit:,.0f}",
            fontsize=12,
            bbox=dict(facecolor="lightgreen",
                      edgecolor="black",
                      boxstyle="round,pad=0.5"))

plt.figtext(0.52,0.90,
            f"Average Discount\n{avg_discount:.2f}%",
            fontsize=12,
            bbox=dict(facecolor="orange",
                      edgecolor="black",
                      boxstyle="round,pad=0.5"))

plt.figtext(0.76,0.90,
            f"Total Orders\n{total_orders}",
            fontsize=12,
            bbox=dict(facecolor="pink",
                      edgecolor="black",
                      boxstyle="round,pad=0.5"))


# Chart 1 - Region Sales

plt.subplot(2,2,1)
sns.barplot(x=region_sales.index,
            y=region_sales.values,
            hue=region_sales.index,
            palette="Set2",
            legend=False)

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")


# Chart 2 - Monthly Sales

plt.subplot(2,2,2)
plt.plot(monthly_sales.index,
         monthly_sales.values,
         marker="o",
         linewidth=2,
         color="blue")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

# Chart 3 - Category Sales

plt.subplot(2,2,3)

explode = [0.1 if i == category_sales.argmax() else 0
           for i in range(len(category_sales))]

plt.pie(category_sales,
        labels=category_sales.index,
        autopct="%1.1f%%",
        explode=explode,
        shadow=True,
        startangle=90)

plt.title("Sales by Category")


# Chart 4 - Discount vs Profit

plt.subplot(2,2,4)

sns.scatterplot(data=df,
                x="Discount",
                y="Profit",
                color="red",
                s=100)

plt.title("Discount vs Profit")
plt.xlabel("Discount (%)")
plt.ylabel("Profit")


# Adjust Layout

plt.tight_layout(rect=[0,0,1,0.87])

# Save Dashboard
plt.savefig("dashboard.png", dpi=300)

# Show Dashboard
plt.show()
```

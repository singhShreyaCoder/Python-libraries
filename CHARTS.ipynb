```python
#Line Chart
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
x1=[2,4,5,6]
y = [10, 20, 25, 30]

plt.plot(x, y, color='green', marker='<', linestyle='-', linewidth=2)
plt.plot(x1, y, color='blue', marker='<', linestyle='--', linewidth=2)
plt.title("Line Chart")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
#plt.legend()
plt.show()
```

```python
#Bar Chart
categories = ['A', 'B', 'C']
values = [20, 30, 40]

plt.bar(categories, values)
plt.title("Bar Plot")
plt.show()

#Set single colour for bars
plt.bar(categories, values, color='red')
plt.title("Bar Plot")
plt.show()

#Different color for each bar
colors = ['red', 'green', 'blue']
plt.bar(categories, values, color=colors)

```

```python
#Basic scattered plot with custom color
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 6, 3, 8, 9, 7, 5]
y = [10, 20, 25, 30, 20, 15, 13, 28, 30, 25]

plt.scatter(x, y, color='purple', s=80)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
#Scattered plot with different colors(value based)
colors = ['red' if val < 15 else 'orange' if val < 25 else 'green' for val in y]

plt.scatter(x, y, color=colors, s=80)
plt.title("Scatter Plot (Conditional Colors)")
plt.show()
```

Histogram


```python
import numpy as np
import matplotlib.pyplot as plt
data = np.random.randn(1000)

plt.hist(data, bins=20, color='skyblue', edgecolor='black')
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
#Histogram with value based colors
fig, ax = plt.subplots()
counts, bins, patches = ax.hist(data, bins=30, edgecolor='black')

for patch, left, right in zip(patches, bins[:-1], bins[1:]):
    if right < -1:
        patch.set_facecolor('orange')
    elif right < 1:
        patch.set_facecolor('white')
    else:
        patch.set_facecolor('green')

ax.set_title("Histogram with Conditional Colors")
ax.set_xlabel("Value")
ax.set_ylabel("Frequency")

plt.show()

```

```python
#Histogram using Color map
fig, ax = plt.subplots()
counts, bins, patches = ax.hist(data, bins=30)

for patch, count in zip(patches, counts):
    patch.set_facecolor(plt.cm.viridis(count / max(counts)))

ax.set_title("Histogram with Colormap")
ax.set_xlabel("Value")
ax.set_ylabel("Frequency")

plt.show()
```

```python
#Histogram with transparency & grid
plt.hist(data, bins=30, color='purple', alpha=0.9)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.title("Histogram with Grid")
plt.show()
```

```python
#PIE CHART
import matplotlib.pyplot as plt
labels=['A','B','C','D']
sizes=[20,30,25,25]
colors=['red','green','blue','orange']

plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    startangle=90,
    autopct='%1.1f%%'#auto percentage
)

plt.title('Pie Chart Example')
plt.show()

#pie chart with exploded slices
explode=(0.1,0.1,0.1,0.1)
plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    startangle=90,
    autopct='%1.1f%%',#auto percentage
    explode=explode,
    shadow=True
)
```

```python
#Area Chart(magnitude of change over time)
import matplotlib.pyplot as plt
import numpy as np
x=[1,2,3,4,5]
y=[10,20,15,25,30]

plt.fill_between(x,y,color="black",alpha=0.5)

plt.plot(x,y,color="blue")
plt.title("area chart")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
#multiple area chart
import matplotlib.pyplot as plt

y1=[10,20,15,25,30]
y2=[5,8,10,15,20]
x=[1,2,3,4,5]
plt.stackplot(x,y1,y2,labels=["y1","y2"],alpha=0.8)

plt.title("stacked area chart")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend(loc="upper left")
plt.show()
```

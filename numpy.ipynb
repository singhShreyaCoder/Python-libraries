```python
import numpy as np

x=np.array([
    [25,5000,60,2],
    [30,6000,75,5],
    [np.nan,55000,68,3],
    [40,np.nan,80,10],
    [35,7000,np.nan,6],
    [28,52000,65,2],
    [45,900000,85,12],
    [32,58000,70,np.nan],
    [50,120000, 95,15],
    [27,51000,62,1]
])

y=np.array([0,1,0,1,1,0,1,0,1,0])

print(x.shape)
print(x.dtype)
print(x.ndim)
print(x.size)
print(np.isnan(x))
print(np.isnan(x).sum(axis=0))
col_means=(np.nanmean(x,axis=0))
print(col_means)
rows,cols=np.where(np.isnan(x))
print(rows,cols)


x[rows,cols] = col_means[cols]
print(x)

print(x[:3])
print(x[6:9])
```

```python
c=np.array([
    [[1,2,4,5],[2,3,4,5],[2,1,3,2]],
    [[3,4,5,6],[1,2,3,4,],[4,5,6,7]]
])

d=np.array([
    [[1,2,3,4],[1,2,4,5],[4,5,6,7]],
    [[1,5,6,7],[5,6,7,8],[6,5,4,3]]
])

print(np.add(c,d))
print(np.subtract(c,d))
print(np.multiply(c,d))
print(np.divide(c,d))
print(np.mod(c,d))
print(np.power(c,d))
```

```python
#scalar product
print(np.add(a,5))
print(np.subtract(a,4))
print(np.multiply(a,5))
print(np.divide(a,5))
print(np.mod(a,3))

print(np.sqrt(a))
print(np.square(a))
print(np.exp(a))
print(np.log(a))
print(np.log10(a))

#trigonometric functions
angles=np.array([0,np.pi/2, np.pi])
print(angles)
print(np.sin(angles))
print(np.cos(angles))
print(np.tan(angles))

#rounding and absolute functions
x=np.array([1.2,2.7,-3.5])
print(np.round(x))
print(np.floor(x))
print(np.ceil(x))
print(np.abs(x))
```

```python
#mean and median

data=np.array([10,20,30,40,50])
print(np.median(data))

#variance and standard deviation
print(np.var(data))
print(np.std(data))

#minimum maximum and range
print(np.min(data))
print(np.max(data))
print(np.ptp(data))

#sum and cummulative sum
print(np.sum(data))
print(np.cumsum(data))

#axis-based operations
arr=np.array([[1,2,3],
              [4,5,6]])

#column wise
print(np.sum(arr,axis=0))
print(np.mean(arr,axis=0))

#row wise
print(np.sum(arr,axis=1))
print(np.mean(arr,axis=1))

#random no function
print(np.random.rand(5))
print(np.random.randn(5))
print(np.random.randint(1,10,5))

#corelation and covariance
x=np.array([1,2,3,4,5])
y=np.array([5,4,3,2,1])
print(np.corrcoef(x,y))
print(np.cov(x,y))

#sorting and indexing  function

arr=np.array([3,1,4,2])
print(np.sort(arr))
print(np.argmax(arr))
print(np.argmin(arr))
```

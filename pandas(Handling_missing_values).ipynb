### New Sample Dataset

Based on your request, I'm creating a new sample dataset with the columns: `name`, `age`, `job`, and `salary`.

```python
import pandas as pd
import numpy as np
from numpy import nan

data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve','Gavin','Utkarsh','Nakul','Shreya'],
    'age': [30, 24, 35, 29, 42,34,20,21, nan],
    'job': ['Engineer', 'Designer', 'Manager', 'Analyst', 'Developer','pilot','Developer','scientist','designer'],
    'salary': [70000, 55000, 90000, 60000, 85000,54000,78000,87000,98000]
}

df = pd.DataFrame(data)
df=df.fillna(0)
print(df)
```

```python
#forwadfill
forward_df=df.fillna(method='ffill')
#print(forward_df)

#backwardfill
backward_df=df.fillna(method='bfill')
print(backward_df)

df=df["name"].fillna("unknown")
print(df)
```

```python
import pandas as pd
import numpy as np
from numpy import nan

data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', None ,'Utkarsh','Nakul','Shreya'],
    'age': [30, 24, 35, 29, 42,34,20,21,nan],
    'job': ['Engineer', 'Designer', 'Manager', 'Analyst', 'Developer','pilot','Developer','scientist','designer'],
    'salary': [70000, 55000, 90000, 60000, 85000,54000,78000,87000,98000]
}

df = pd.DataFrame(data)

num_cols=df.select_dtypes(include=["float64","int64"]).columns
for col in num_cols:
  df[col]=df[col].fillna(df[col].mean())
print(df)
```

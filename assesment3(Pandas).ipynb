```python
#Question 1DataFrame Creation and Filtering

import pandas as pd

data = {
    "Emp_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Department": ["HR", "IT", "Finance", "IT", "HR"],
    "Salary": [45000, 65000, 55000, 70000, 48000],
    "Experience": [3, 6, 5, 8, 4]
}

df = pd.DataFrame(data)
# 1. Employees from IT department
print(df[df["Department"] == "IT"])

#2. Employees with Salary greater than 50000
print(df[df["Salary"] > 50000])

# 3. Add Bonus column (10% of Salary)
df["Bonus"] = df["Salary"] * 0.10

#4.Sort the DataFrame by Salary in descending order.
sorted_df=df.sort_values(by="Salary",ascending=False)
print(sorted_df)

#5. show only name salary and bonus
print(df[["Name","Salary","Bonus"]])

```

```python
#Question 2 Missing Values Analysis and Handling

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Student": ["A", "B", "C", None, "E"],
    "Math": [85, np.nan, 76, 92, np.nan],
    "Science": [90, 78, np.nan, 89, 85],
    "English": [88, 82, 80, np.nan, 91]
})

# 1. Total missing values
missing_values = df.isnull().sum().sum()
print("Total missing value:",missing_values)


# 2. Missing values column-wise
print(df.isnull().sum())

# 3. Fill missing values in Math with median
df["Math"] = df["Math"].fillna(df["Math"].median())
#print(df)

# 4. Fill missing values in Science with mean
df["Science"] = df["Science"].fillna(df["Science"].mean())
#print(df)

# 5. Fill missing values in English using forward fill
df["English"] = df["English"].ffill()
#print(df)

#5.fill missing value with mode
df["Student"] = df["Student"].fillna(df["Student"].mode())
print(df)

```

```python
#Question 3. Duplicate Records and Data Cleaning

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Order_ID": [201,202,203,202,204,201],
    "Customer":['john','sara','john','sara','mike','john'],
    "Product" :['laptop','mouse','keyboard','mouse','monitor','laptop'],
    "Amount"  :[65000,800,1500,800,12000,65000]
})
#1. identify dublicate rows
duplicate_rows=df[df.duplicated()]
print(duplicate_rows)

#2.Count duplicate records
duplicate_rows=df.duplicated().sum().sum()
print(duplicate_rows)

#3.Remove duplicate rows while keeping the first occurrence.
df.drop_duplicates(inplace=True)
print(df)

#4.Reset the index after removing duplicates.
df.reset_index(drop=True,inplace=True)
print(df)

#5.Calculate the total sales amount after cleaning the data.
total_sales=df["Amount"].sum()
print(total_sales)

```

```python
#Question 4. Descriptive Statistics and Grouping

import pandas as pd

# Create the DataFrame
data = {
    "Employee": ["A", "B", "C", "D", "E", "F"],
    "Department": ["HR", "IT", "HR", "IT", "Finance", "Finance"],
    "Salary": [45000, 65000, 48000, 72000, 58000, 62000],
    "Age": [25, 30, 27, 35, 29, 32]
}

df = pd.DataFrame(data)

# Display the DataFrame
print("Employee DataFrame:")
print(df)

# 1. Display descriptive statistics for all numeric columns
print("\nDescriptive Statistics:")
print(df.describe())

# 2. Find the average salary of each department
print("\nAverage Salary of Each Department:")
print(df.groupby("Department")["Salary"].mean())

# 3. Find the highest salary in each department
print("\nHighest Salary in Each Department:")
print(df.groupby("Department")["Salary"].max())

# 4. Find the number of employees in each department
print("\nNumber of Employees in Each Department:")
print(df.groupby("Department")["Employee"].count())

# 5. Display employees whose salary is above the overall average salary
overall_avg_salary = df["Salary"].mean()

print("\nOverall Average Salary:", overall_avg_salary)
print("\nEmployees with Salary Above Overall Average:")
print(df[df["Salary"] > overall_avg_salary])
```

```python
#Question 5. Pandas Series Operations
import pandas as pd

# Create the two Series
sales = pd.Series([1200, 1500, 1800, 1700, 2000])
expenses = pd.Series([800, 900, 1200, 1000, 1400])

# Display the Series
print("Sales:")
print(sales)

print("\nExpenses:")
print(expenses)

# Calculate the profit for each entry
profit = sales - expenses
print("\nProfit:")
print(profit)

# Find the maximum and minimum profit
print("\nMaximum Profit:", profit.max())
print("Minimum Profit:", profit.min())

# Display only those profits greater than 600
print("\nProfits greater than 600:")
print(profit[profit > 600])

# Calculate the average profit
print("\nAverage Profit:", profit.mean())
```

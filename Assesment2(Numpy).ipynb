```python
#Question 1: Student Marks Analysis

import numpy as np

# Student IDs
student_ids = np.array(["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"])

# Marks
marks = np.array([78, 85, 92, 67, 88, 45, 73, 95, 81, 60])

# 1. Find the average marks
average_marks = np.mean(marks)
print("Average Marks:", average_marks)

# 2. Find the highest and lowest marks
highest_marks = np.max(marks)
lowest_marks = np.min(marks)
print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)

# 3. Find students scoring above 80
above_80 = student_ids[marks > 80]
print("Students scoring above 80:", above_80)

# 4. Find students scoring below 50
below_50 = student_ids[marks < 50]
print("Students scoring below 50:", below_50)

# 5. Add 5 grace marks to every student using vectorized operations
grace_marks = marks + 5
print("Marks after adding 5 grace marks:", grace_marks)
```

```python
#Question 2: Employee Salary Management
import numpy as np

employee_ids = np.array(["E1", "E2", "E3", "E4", "E5", "E6", "E7", "E8"])
salaries = np.array([25000, 32000, 45000, 55000, 28000, 60000, 75000, 50000])

# 1. Find the average salary
average_salary = np.mean(salaries)
print("Average Salary: ₹", average_salary)

# 2. Find employees earning more than ₹50,000
high_salary = employee_ids[salaries > 50000]
print("Employees earning more than ₹50,000:", high_salary)

# 3. Increase every salary by 10%
updated_salaries = salaries * 1.10
print("Salaries after 10% increase:", updated_salaries)

# 4. Find the total salary expenditure
total_expenditure = np.sum(salaries)
print("Total Salary Expenditure: ₹", total_expenditure)

# 5. Find salaries between ₹30,000 and ₹60,000
salary_range = salaries[(salaries >= 30000) & (salaries <= 60000)]
print("Salaries between ₹30,000 and ₹60,000:", salary_range)
```

```python
#Question 3: E-Commerce Sales Dataset
import numpy as np


sales = np.array([
    [1200, 1500, 1800, 1700, 1600, 2000, 2200],
    [1400, 1600, 1900, 2100, 1800, 2300, 2500]
])

days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])

# 1. Find total sales for each week
total_sales = np.sum(sales, axis=1)
print("Total Sales for Each Week:", total_sales)

# 2. Find average daily sales
average_daily_sales = np.mean(sales, axis=0)
print("Average Daily Sales:", average_daily_sales)

# 3. Find the day with maximum sales
day_index = np.argmax(np.sum(sales, axis=0))
print("Day with Maximum Sales:", days[day_index])

# 4. Extract sales of Week 1 only
week1_sales = sales[0]
print("Week 1 Sales:", week1_sales)

# 5. Extract sales of the weekend (last two columns)
weekend_sales = sales[:, -2:]
print("Weekend Sales:\n", weekend_sales)
```

```python
#Question 4: Hospital Patient Age Analysis
import numpy as np

# Patient IDs
patient_ids = np.array(["P1", "P2", "P3", "P4", "P5", "P6",
                        "P7", "P8", "P9", "P10", "P11", "P12"])

# Ages
ages = np.array([12, 18, 25, 30, 45, 60, 72, 15, 34, 50, 28, 67])

# 1. Find the average age
average_age = np.mean(ages)
print("Average Age:", average_age)

# 2. Find all senior citizens (age >= 60)
senior_citizens = patient_ids[ages >= 60]
print("Senior Citizens:", senior_citizens)

# 3. Find all children (age < 18)
children = patient_ids[ages < 18]
print("Children:", children)

# 4. Count the number of adults (18–59)
adults_count = np.sum((ages >= 18) & (ages <= 59))
print("Number of Adults:", adults_count)

# 5. Replace ages less than 18 with 0
updated_ages = np.where(ages < 18, 0, ages)
print("Updated Ages:", updated_ages)
```

```python
#Question 6: Customer Dataset Preprocessing
import numpy as np

# Customer IDs
customer_ids = np.array(["C1", "C2", "C3", "C4", "C5", "C6"])


data = np.array([
    [25, 50000, 60],
    [30, 60000, 75],
    [np.nan, 55000, 68],
    [40, np.nan, 80],
    [35, 70000, np.nan],
    [28, 52000, 65]
], dtype=float)


# Part A: Data Cleaning

# 1. Count missing values
missing_values = np.isnan(data).sum()
print("Total Missing Values:", missing_values)

# Missing values column-wise
print("Missing Values in Each Column:", np.isnan(data).sum(axis=0))

# 2. Find column means (ignoring NaN values)
column_means = np.nanmean(data, axis=0)
print("Column Means:", column_means)

# 3. Replace missing values with column means
cleaned_data = data.copy()

# Get indices of missing values
rows, cols = np.where(np.isnan(cleaned_data))

# Replace NaN with corresponding column mean
cleaned_data[rows, cols] = column_means[cols]

# 4. Display the cleaned dataset
print("\nCleaned Dataset:")
print(cleaned_data)
```

```python
#Q7 Online Store Customer Segmentation
import numpy as np

# Customer IDs
customer_ids = np.array(["C1", "C2", "C3", "C4", "C5", "C6"])

data = np.array([
    [22, 25000],
    [35, 55000],
    [45, 85000],
    [28, 30000],
    [50, 120000],
    [40, 75000]
], dtype=float)


# 1. Find customers earning > ₹70,000
high_income = customer_ids[data[:, 1] > 70000]
print("Customers earning more than ₹70,000:", high_income)


# 2. Find customers aged between 30 and 50

age_group = customer_ids[(data[:, 0] >= 30) & (data[:, 0] <= 50)]
print("Customers aged between 30 and 50:", age_group)


# 3. Normalize the dataset (Min-Max Scaling)
min_values = np.min(data, axis=0)
max_values = np.max(data, axis=0)

normalized_data = (data - min_values) / (max_values - min_values)

print("\nNormalized Dataset:")
print(normalized_data)


# 4. Standardize the dataset (Z-score)

mean = np.mean(data, axis=0)
std = np.std(data, axis=0)

standardized_data = (data - mean) / std

print("\nStandardized Dataset:")
print(standardized_data)

# Verify Standardization
print("\nMean of Standardized Data:")
print(np.mean(standardized_data, axis=0))

print("\nStandard Deviation of Standardized Data:")
print(np.std(standardized_data, axis=0))


# 5. Shuffle the dataset randomly

np.random.seed(42)  # For reproducible results
shuffled_data = data.copy()
np.random.shuffle(shuffled_data)

print("\nShuffled Dataset:")
print(shuffled_data)


# 6. Split into 80% Training and 20% Testing Data

split_index = int(0.8 * len(shuffled_data))

train_data = shuffled_data[:split_index]
test_data = shuffled_data[split_index:]

print("\nTraining Data:")
print(train_data)

print("\nTesting Data:")
print(test_data)
```
